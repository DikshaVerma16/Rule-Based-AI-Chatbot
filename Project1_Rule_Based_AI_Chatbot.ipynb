Objective

To build a simple AI chatbot using Python and if-else statements that can answer predefined questions.

```python
from datetime import datetime

print("=" * 50)
print("🤖 Welcome to the Rule-Based AI Chatbot!")
print("I can answer some simple questions.")
print("Type 'exit' or 'bye' to end the chat.")
print("=" * 50)

while True:

    user = input("\nYou: ").strip().lower()

    if user == "hello" or user == "hi" or user == "hey":
        print("Bot: Hello! Nice to meet you.")

    elif user == "how are you":
        print("Bot: I am fine. Thanks for asking!")

    elif user == "what is your name" or user == "your name":
        print("Bot: My name is Rule-Based AI Chatbot.")

    elif user == "who created you":
        print("Bot: I was created using Python.")

    elif user == "what is ai":
        print("Bot: AI stands for Artificial Intelligence.")

    elif user == "what is python":
        print("Bot: Python is a popular programming language used for AI, web development, and data science.")

    elif user == "good morning":
        print("Bot: Good Morning! Have a wonderful day.")

    elif user == "good afternoon":
        print("Bot: Good Afternoon!")

    elif user == "good evening":
        print("Bot: Good Evening!")

    elif user == "good night":
        print("Bot: Good Night! Sweet dreams.")

    elif user == "thank you" or user == "thanks":
        print("Bot: You're welcome!")

    elif user == "help":
        print("Bot: You can ask me:")
        print("- hello")
        print("- how are you")
        print("- what is ai")
        print("- what is python")
        print("- what is your name")
        print("- who created you")
        print("- date")
        print("- time")
        print("- bye")

    elif user == "date":
        print("Bot:", datetime.now().strftime("%d-%m-%Y"))

    elif user == "time":
        print("Bot:", datetime.now().strftime("%I:%M:%S %p"))

    elif user == "bye" or user == "exit":
        print("Bot: Goodbye! Have a nice day.")
        break

    else:
        print("Bot: Sorry, I don't understand that. Type 'help' to see what I can answer.")
```

Conclusion

This project demonstrates the implementation of a simple rule-based AI chatbot using Python. The chatbot responds to predefined user inputs using conditional statements and continues running until the user exits.
